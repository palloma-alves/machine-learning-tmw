Na árvore de decisão o objetivo é buscar nós puros.

É feito a divisão de acordo com o grau de pureza, buscando nós mais puros.

Nó puro é quando só tem dados de uma categoria.

Quanto mais fundo a árvore vai ser, mais puro os nós serão, no entanto terá menos amostras nos nós.
